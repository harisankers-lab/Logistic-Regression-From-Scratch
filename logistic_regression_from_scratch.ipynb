```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression as lr_model
from sklearn.linear_model import LogisticRegression as lg_model
```

```python
class LinearRegression:

  def fit(self, X, y):

    x_mean = np.mean(X)
    y_mean = np.mean(y)

    numerator = np.sum((X - x_mean) * (y - y_mean))
    denominator = np.sum((X - x_mean) ** 2)

    self.m = numerator / denominator
    self.c = y_mean - (self.m * x_mean)

  def predict(self, x):
    y_pred = (self.m * x) + self.c
    return y_pred
```

```python
class LogisticRegression:

  def fit(self, X, y):
    self.m = 0
    self.c = 0

    learning_rate = 0.01
    epochs = 10000
    n = len(X)

    for i in range(epochs):
      y_pred = self.m * X + self.c
      probability = 1 / (1 + np.exp(-y_pred))

      error = probability - y

      grad_m = np.sum(error * X) / n
      grad_c = np.sum(error) / n

      self.m = self.m - learning_rate * grad_m
      self.c = self.c - learning_rate * grad_c

  def predict(self, x):
    y_pred = (self.m * x) + self.c
    probability = 1 / (1 + np.exp(-y_pred))
    threshold = 0.5
    prediction = 1 if probability >= threshold else 0
    return (probability, prediction)
```

```python
df = pd.DataFrame(
    {
        "car_years" : [1, 2, 2, 3, 4, 6, 8, 10, 12, 15],
        "resale_price" : [20.5, 19.5, 19.2, 18.3, 17.5, 16.4, 13.9, 11.0, 7.8, 4.2]
    }
)

X = df["car_years"]
y = df["resale_price"]
```

```python
linear_model = LinearRegression()
linear_model.fit(X, y)

x = 7
prediction = linear_model.predict(x)

print("Our Linear Regression")
print(f"{"-"*30}")
print(f"No. of years: {x}\nPrediction: {prediction} lakhs")

linear_reg  = lr_model()
linear_reg.fit(df[["car_years"]], y)

linear_reg_prediction = linear_reg.predict(
    pd.DataFrame({"car_years": [7]})
)

print("")
print("Sklearn Linear Regression")
print(f"{"-"*30}")
print("Prediction:", linear_reg_prediction[0], "lakhs")
```

```python
y_class = (df["resale_price"] > 15).astype(int)

log_model = LogisticRegression()
log_model.fit(X, y_class)

x = 10
prob, log_pred = log_model.predict(x)

print("Our Logistic Regression")
print(f"{"-"*30}")
print(f"No of years: {x}\nPrediction class: {log_pred}")

if log_pred == 1:
  print("Expensive")
else:
  print("Affordable")


log_reg = lg_model()

log_reg.fit(df[["car_years"]], y_class)

log_test = pd.DataFrame({"car_years": [x]})

log_class = log_reg.predict(log_test)[0]

print(" ")
print("Sklearn Logistic Regression")
print(f"{"-"*30}")
print("Predicted class:", log_class)
```
